```python
import pandas as pd
from scipy.stats import wilcoxon
```

# Case, where each thread has <$n:=$ number of threads> tasks

```python
thread_pool_1 = pd.read_csv("./integral_parallel_measurements/measurements_thread_pool_1_partitions.csv")
work_stealing_1 = pd.read_csv("./integral_parallel_measurements/measurements_stealing_1_partitions.csv")
```

```python
columns = ["1", "2", "3"]
results = {}

for col in columns:
    stat, pvalue = wilcoxon(thread_pool_1[col], work_stealing_1[col])
    results[col] = stat, pvalue

for i, values in results.items():
    stats, pvalue = values
    print(f"Column: {i}: Statistics {stats}, P-value: {pvalue}")
```

As in all the cases p-value of the test is very high, we cannot reject two-sided wilcoxon test's null hypothesis (it states that time efficiency of these two approaches should be almost the same).

# Each thread has $3n$ tasks

```python
thread_pool_3 = pd.read_csv("./integral_parallel_measurements/measurements_thread_pool_3_partitions.csv")
work_stealing_3 = pd.read_csv("./integral_parallel_measurements/measurements_stealing_3_partitions.csv")
```

```python

for col in columns:
    stat, pvalue = wilcoxon(thread_pool_3[col], work_stealing_3[col])
    results[col] = stat, pvalue

for i, values in results.items():
    stats, pvalue = values
    print(f"Column: {i}: Statistics {stats}, P-value: {pvalue}")
```

The same situation is noted, when we assign 3*threads tasks to each worker. However, p-value is much smaller. For third function, if we set size of the test to be 0.1, we even can say that work stealing approach is better.

# Each thread has $8n$ tasks

```python
thread_pool_8 = pd.read_csv("./integral_parallel_measurements/measurements_thread_pool_8_partitions.csv")
work_stealing_8 = pd.read_csv("./integral_parallel_measurements/measurements_stealing_8_partitions.csv")


for col in columns:
    stat, pvalue = wilcoxon(thread_pool_8[col], work_stealing_8[col])
    results[col] = stat, pvalue

for i, values in results.items():
    stats, pvalue = values
    print(f"Column: {i}: Statistics {stats}, P-value: {pvalue}")
```

When each thread has substantial amount of jobs, we can see that pvalues are lower than standard size of test 0.05. Test's statistics are also positive, thus we conclude that indeed work stealing approach gives us boost in performance, when we may meet the situation that some worker does not have a job.
